## 🧠 Integrated System Classes

Simplified versions of all system components for integration testing.

```python
class DatingEngineSystem:
    """Master system integrating all dating engine components"""
    
    def __init__(self, config: Dict[str, Any] = None):
        self.config = config or SYSTEM_CONFIG
        self.users_db = []  # In-memory user database
        self.questions_db = []  # Question bank
        self.compatibility_cache = {}  # Cache for compatibility scores
        
        # Initialize components
        self.reddit_collector = RedditDataCollector()
        self.mbti_engine = MBTIEngine()
        self.compatibility_engine = CompatibilityEngine(self.config['weights'])
        self.matcher = UserMatcher(self.config['thresholds'])
        
        print(f'🚀 {self.config["name"]} initialized successfully!')
    
    def load_sample_data(self) -> Dict[str, int]:
        """Load sample data for testing"""
        print('📊 Loading sample data...')
        
        # Sample questions
        self.questions_db = [
            {'id': 'q1', 'text': 'What do you value most in a relationship?', 'category': 'values'},
            {'id': 'q2', 'text': 'How do you prefer to spend your weekends?', 'category': 'lifestyle'},
            {'id': 'q3', 'text': 'What is your communication style?', 'category': 'communication'},
            {'id': 'q4', 'text': 'How do you handle conflicts?', 'category': 'conflict_resolution'},
            {'id': 'q5', 'text': 'What are your long-term goals?', 'category': 'future_planning'}
        ]
        
        # Sample users
        sample_users = self._generate_sample_users(50)
        self.users_db.extend(sample_users)
        
        return {
            'users': len(self.users_db),
            'questions': len(self.questions_db)
        }
    
    def _generate_sample_users(self, count: int) -> List[Dict[str, Any]]:
        """Generate sample users for testing"""
        users = []
        mbti_types = ['INTJ', 'ENFP', 'ISTJ', 'ESFP', 'INFJ', 'ENTP', 'ISFJ', 'ESTP']
        zodiac_signs = ['Aries', 'Taurus', 'Gemini', 'Cancer', 'Leo', 'Virgo']
        communication_styles = ['Direct', 'Indirect', 'Analytical', 'Supportive']
        
        sample_answers = {
            'q1': ['Trust and honesty', 'Fun and adventure', 'Stability and security', 'Growth and learning'],
            'q2': ['Reading and relaxing', 'Outdoor activities', 'Social gatherings', 'Creative projects'],
            'q3': ['Direct and clear', 'Thoughtful and careful', 'Emotional and expressive', 'Logical and structured'],
            'q4': ['Address immediately', 'Take time to think', 'Seek compromise', 'Avoid confrontation'],
            'q5': ['Career advancement', 'Family and relationships', 'Personal growth', 'Financial security']
        }
        
        for i in range(count):
            # Generate answers for each question
            answers = {}
            for q_id, options in sample_answers.items():
                answers[q_id] = random.choice(options)
            
            user = {
                'user_id': f'user_{i+1:03d}',
                'name': f'User {i+1}',
                'age': random.randint(22, 45),
                'gender': random.choice(['M', 'F']),
                'mbti_type': random.choice(mbti_types),
                'astrology_sign': random.choice(zodiac_signs),
                'communication_style': random.choice(communication_styles),
                'interests': random.sample(['music', 'travel', 'cooking', 'sports', 'reading', 'movies', 'art', 'technology'], k=random.randint(3, 6)),
                'prompt_answers': answers,
                'created_date': datetime.now().isoformat()
            }
            users.append(user)
        
        return users
    
    def find_matches(self, user_id: str, limit: int = 10) -> List[Dict[str, Any]]:
        """Find compatible matches for a user"""
        target_user = next((u for u in self.users_db if u['user_id'] == user_id), None)
        if not target_user:
            return []
        
        matches = []
        for candidate in self.users_db:
            if candidate['user_id'] != user_id and candidate['gender'] != target_user['gender']:
                score = self.compatibility_engine.calculate_compatibility(target_user, candidate)
                matches.append({
                    'user_id': candidate['user_id'],
                    'name': candidate['name'],
                    'compatibility_score': score,
                    'mbti_type': candidate['mbti_type'],
                    'age': candidate['age'],
                    'interests': candidate['interests']
                })
        
        # Sort by compatibility score and return top matches
        matches.sort(key=lambda x: x['compatibility_score'], reverse=True)
        return matches[:limit]
    
    def generate_system_report(self) -> Dict[str, Any]:
        """Generate comprehensive system performance report"""
        print('📊 Generating system performance report...')
        
        # Calculate basic statistics
        total_users = len(self.users_db)
        gender_distribution = Counter(u['gender'] for u in self.users_db)
        mbti_distribution = Counter(u['mbti_type'] for u in self.users_db)
        age_stats = {
            'min': min(u['age'] for u in self.users_db) if self.users_db else 0,
            'max': max(u['age'] for u in self.users_db) if self.users_db else 0,
            'avg': sum(u['age'] for u in self.users_db) / len(self.users_db) if self.users_db else 0
        }
        
        # Test matching performance
        if self.users_db:
            sample_user = self.users_db[0]
            start_time = time.time()
            matches = self.find_matches(sample_user['user_id'], limit=5)
            matching_time = time.time() - start_time
            avg_compatibility = sum(m['compatibility_score'] for m in matches) / len(matches) if matches else 0
        else:
            matching_time = 0
            avg_compatibility = 0
        
        report = {
            'timestamp': datetime.now().isoformat(),
            'system_info': self.config,
            'data_statistics': {
                'total_users': total_users,
                'total_questions': len(self.questions_db),
                'gender_distribution': dict(gender_distribution),
                'mbti_distribution': dict(mbti_distribution),
                'age_statistics': age_stats
            },
            'performance_metrics': {
                'matching_time_seconds': matching_time,
                'average_compatibility_score': avg_compatibility,
                'cache_size': len(self.compatibility_cache)
            },
            'system_health': {
                'status': 'operational',
                'components_loaded': len(self.config['components']),
                'data_integrity': 'verified'
            }
        }
        
        return report

# Supporting classes (simplified versions)
class RedditDataCollector:
    def collect_questions(self): return {'status': 'simulated', 'questions_collected': 100}

class MBTIEngine:
    def predict_type(self, text): return ('INTJ', 0.85)
    def get_compatibility(self, type1, type2): return random.uniform(60, 95)

class CompatibilityEngine:
    def __init__(self, weights): self.weights = weights
    def calculate_compatibility(self, user1, user2):
        # Simplified compatibility calculation
        mbti_score = random.uniform(60, 95) if user1['mbti_type'] != user2['mbti_type'] else 85
        interest_overlap = len(set(user1['interests']).intersection(set(user2['interests']))) / len(set(user1['interests']).union(set(user2['interests']))) * 100
        prompt_score = random.uniform(50, 90)
        return (mbti_score * 0.4 + interest_overlap * 0.3 + prompt_score * 0.3)

class UserMatcher:
    def __init__(self, thresholds): self.thresholds = thresholds
    def classify_match(self, score):
        if score >= self.thresholds['high_compatibility']: return 'high'
        elif score >= self.thresholds['medium_compatibility']: return 'medium'
        else: return 'low'

print('🧠 Integrated system classes defined successfully!')
```
