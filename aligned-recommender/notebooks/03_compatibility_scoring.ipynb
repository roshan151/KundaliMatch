## 🔧 Configuration & Setup

Setting up compatibility scoring weights and parameters.

```python
# Project directories
BASE_DIR = Path('../')
OUTPUTS_DIR = BASE_DIR / 'outputs' / 'csv'
MODELS_DIR = BASE_DIR / 'models'

# Create directories
for directory in [OUTPUTS_DIR, MODELS_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

# Compatibility scoring weights (must sum to 1.0)
SCORING_WEIGHTS = {
    'mbti': 0.30,        # MBTI personality compatibility
    'prompts': 0.30,     # Response similarity to prompts
    'interests': 0.20,   # Shared interests and hobbies
    'communication': 0.10,  # Communication style compatibility
    'astrology': 0.10    # Astrological compatibility
}

# Validate weights
total_weight = sum(SCORING_WEIGHTS.values())
if abs(total_weight - 1.0) > 0.001:
    print(f"⚠️ Warning: Weights sum to {total_weight}, not 1.0")
else:
    print(f"✅ Scoring weights configured: {SCORING_WEIGHTS}")

print(f"📁 Output directory: {OUTPUTS_DIR}")
print(f"🤖 Models directory: {MODELS_DIR}")
```

## 🧠 AI-Powered Semantic Similarity

Using sentence transformers for semantic analysis of user responses.

```python
# Initialize sentence transformer model
try:
    print("🤖 Loading sentence transformer model...")
    sentence_model = SentenceTransformer('all-MiniLM-L6-v2')
    print("✅ Sentence transformer loaded successfully!")
    ai_available = True
except Exception as e:
    print(f"⚠️ Could not load sentence transformer: {e}")
    print("📝 Using fallback similarity calculation")
    sentence_model = None
    ai_available = False

def calculate_semantic_similarity(text1: str, text2: str) -> float:
    """Calculate semantic similarity between two texts"""
    if ai_available and sentence_model:
        # Use AI-powered semantic similarity
        embeddings = sentence_model.encode([text1, text2])
        similarity = cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]
        return float(similarity)
    else:
        # Fallback: simple word overlap similarity
        words1 = set(text1.lower().split())
        words2 = set(text2.lower().split())
        intersection = len(words1.intersection(words2))
        union = len(words1.union(words2))
        return intersection / union if union > 0 else 0.0

# Test semantic similarity
test_texts = [
    "I love outdoor activities and hiking",
    "I enjoy nature walks and mountain climbing",
    "I prefer staying indoors and reading books"
]

print("\n🧪 Testing semantic similarity:")
for i in range(len(test_texts)):
    for j in range(i+1, len(test_texts)):
        similarity = calculate_semantic_similarity(test_texts[i], test_texts[j])
        print(f"   Text {i+1} vs Text {j+1}: {similarity:.3f}")
```

## 🎯 Compatibility Scoring Engines

Individual scoring engines for each compatibility dimension.

```python
class CompatibilityScorer:
    def __init__(self, weights: Dict[str, float]):
        self.weights = weights
        self.mbti_compatibility = self._create_mbti_matrix()
        self.astrology_compatibility = self._create_astrology_matrix()
    
    def _create_mbti_matrix(self) -> Dict[Tuple[str, str], float]:
        """Create MBTI compatibility matrix"""
        mbti_types = ['INTJ', 'INTP', 'ENTJ', 'ENTP', 'INFJ', 'INFP', 'ENFJ', 'ENFP',
                      'ISTJ', 'ISFJ', 'ESTJ', 'ESFJ', 'ISTP', 'ISFP', 'ESTP', 'ESFP']
        
        # Golden pairs (highest compatibility)
        golden_pairs = {
            'INTJ': ['ENFP', 'ENTP'], 'INTP': ['ENFJ', 'ENTJ'],
            'ENTJ': ['INFP', 'INTP'], 'ENTP': ['INFJ', 'INTJ'],
            'INFJ': ['ENFP', 'ENTP'], 'INFP': ['ENFJ', 'ENTJ'],
            'ENFJ': ['INFP', 'INTP'], 'ENFP': ['INFJ', 'INTJ'],
            'ISTJ': ['ESFP', 'ESTP'], 'ISFJ': ['ESFP', 'ESTP'],
            'ESTJ': ['ISFP', 'ISTP'], 'ESFJ': ['ISFP', 'ISTP'],
            'ISTP': ['ESFJ', 'ESTJ'], 'ISFP': ['ESFJ', 'ESTJ'],
            'ESTP': ['ISFJ', 'ISTJ'], 'ESFP': ['ISFJ', 'ISTJ']
        }
        
        compatibility = {}
        for type1 in mbti_types:
            for type2 in mbti_types:
                if type2 in golden_pairs.get(type1, []):
                    score = random.uniform(0.85, 0.95)
                elif type1 == type2:
                    score = random.uniform(0.75, 0.85)
                else:
                    score = random.uniform(0.60, 0.80)
                compatibility[(type1, type2)] = score
        
        return compatibility
    
    def _create_astrology_matrix(self) -> Dict[Tuple[str, str], float]:
        """Create astrology compatibility matrix"""
        signs = ['Aries', 'Taurus', 'Gemini', 'Cancer', 'Leo', 'Virgo',
                'Libra', 'Scorpio', 'Sagittarius', 'Capricorn', 'Aquarius', 'Pisces']
        
        # Element compatibility
        elements = {
            'Fire': ['Aries', 'Leo', 'Sagittarius'],
            'Earth': ['Taurus', 'Virgo', 'Capricorn'],
            'Air': ['Gemini', 'Libra', 'Aquarius'],
            'Water': ['Cancer', 'Scorpio', 'Pisces']
        }
        
        def get_element(sign):
            for element, sign_list in elements.items():
                if sign in sign_list:
                    return element
            return None
        
        compatibility = {}
        for sign1 in signs:
            for sign2 in signs:
                element1 = get_element(sign1)
                element2 = get_element(sign2)
                
                if element1 == element2:
                    score = random.uniform(0.80, 0.90)
                elif (element1 in ['Fire', 'Air'] and element2 in ['Fire', 'Air']) or \
                     (element1 in ['Earth', 'Water'] and element2 in ['Earth', 'Water']):
                    score = random.uniform(0.70, 0.85)
                else:
                    score = random.uniform(0.50, 0.75)
                
                compatibility[(sign1, sign2)] = score
        
        return compatibility
    
    def score_mbti_compatibility(self, type1: str, type2: str) -> float:
        """Score MBTI compatibility between two types"""
        return self.mbti_compatibility.get((type1, type2), 0.65)
    
    def score_astrology_compatibility(self, sign1: str, sign2: str) -> float:
        """Score astrological compatibility between two signs"""
        return self.astrology_compatibility.get((sign1, sign2), 0.65)
    
    def score_interests_compatibility(self, interests1: List[str], interests2: List[str]) -> float:
        """Score compatibility based on shared interests"""
        if not interests1 or not interests2:
            return 0.5
        
        set1 = set(interests1)
        set2 = set(interests2)
        
        intersection = len(set1.intersection(set2))
        union = len(set1.union(set2))
        
        jaccard_similarity = intersection / union if union > 0 else 0
        
        # Scale to 0.3-1.0 range (some compatibility even with no shared interests)
        return 0.3 + (jaccard_similarity * 0.7)
    
    def score_prompt_compatibility(self, answers1: Dict[str, str], answers2: Dict[str, str]) -> float:
        """Score compatibility based on prompt responses"""
        if not answers1 or not answers2:
            return 0.5
        
        common_questions = set(answers1.keys()).intersection(set(answers2.keys()))
        if not common_questions:
            return 0.5
        
        similarities = []
        for question in common_questions:
            similarity = calculate_semantic_similarity(answers1[question], answers2[question])
            similarities.append(similarity)
        
        return np.mean(similarities) if similarities else 0.5
    
    def score_communication_compatibility(self, style1: str, style2: str) -> float:
        """Score communication style compatibility"""
        # Communication style compatibility matrix
        compatibility_map = {
            ('Direct', 'Direct'): 0.85,
            ('Direct', 'Indirect'): 0.60,
            ('Direct', 'Analytical'): 0.75,
            ('Direct', 'Supportive'): 0.70,
            ('Indirect', 'Indirect'): 0.80,
            ('Indirect', 'Analytical'): 0.65,
            ('Indirect', 'Supportive'): 0.85,
            ('Analytical', 'Analytical'): 0.90,
            ('Analytical', 'Supportive'): 0.70,
            ('Supportive', 'Supportive'): 0.85
        }
        
        # Make symmetric
        score = compatibility_map.get((style1, style2)) or compatibility_map.get((style2, style1))
        return score if score is not None else 0.65
    
    def calculate_overall_compatibility(self, user1: Dict[str, Any], user2: Dict[str, Any]) -> Dict[str, float]:
        """Calculate overall compatibility score between two users"""
        scores = {}
        
        # MBTI compatibility
        scores['mbti'] = self.score_mbti_compatibility(
            user1.get('mbti_type', 'INTJ'),
            user2.get('mbti_type', 'ENFP')
        )
        
        # Astrology compatibility
        scores['astrology'] = self.score_astrology_compatibility(
            user1.get('astrology_sign', 'Aries'),
            user2.get('astrology_sign', 'Leo')
        )
        
        # Interests compatibility
        scores['interests'] = self.score_interests_compatibility(
            user1.get('interests', []),
            user2.get('interests', [])
        )
        
        # Prompt responses compatibility
        scores['prompts'] = self.score_prompt_compatibility(
            user1.get('prompt_answers', {}),
            user2.get('prompt_answers', {})
        )
        
        # Communication style compatibility
        scores['communication'] = self.score_communication_compatibility(
            user1.get('communication_style', 'Direct'),
            user2.get('communication_style', 'Supportive')
        )
        
        # Calculate weighted overall score
        overall_score = sum(scores[factor] * self.weights[factor] for factor in scores)
        scores['overall'] = overall_score
        
        return scores

# Initialize the compatibility scorer
compatibility_scorer = CompatibilityScorer(SCORING_WEIGHTS)
print("💕 Compatibility scorer initialized successfully!")
print(f"   MBTI pairs: {len(compatibility_scorer.mbti_compatibility)}")
print(f"   Astrology pairs: {len(compatibility_scorer.astrology_compatibility)}")
```

## 🧪 Testing Compatibility Scoring

Testing the compatibility scoring system with sample user profiles.

```python
# Create sample user profiles for testing
sample_users = [
    {
        'user_id': 'user_001',
        'name': 'Alice',
        'mbti_type': 'INTJ',
        'astrology_sign': 'Scorpio',
        'interests': ['reading', 'technology', 'chess', 'philosophy'],
        'communication_style': 'Analytical',
        'prompt_answers': {
            'q1': 'I value deep, meaningful conversations over small talk',
            'q2': 'I prefer planning ahead and having a clear structure',
            'q3': 'I enjoy intellectual challenges and problem-solving'
        }
    },
    {
        'user_id': 'user_002',
        'name': 'Bob',
        'mbti_type': 'ENFP',
        'astrology_sign': 'Gemini',
        'interests': ['travel', 'music', 'art', 'meeting new people'],
        'communication_style': 'Supportive',
        'prompt_answers': {
            'q1': 'I love connecting with people and sharing experiences',
            'q2': 'I prefer to go with the flow and adapt as needed',
            'q3': 'I enjoy creative pursuits and exploring new possibilities'
        }
    },
    {
        'user_id': 'user_003',
        'name': 'Carol',
        'mbti_type': 'ISFJ',
        'astrology_sign': 'Cancer',
        'interests': ['cooking', 'gardening', 'family time', 'volunteering'],
        'communication_style': 'Supportive',
        'prompt_answers': {
            'q1': 'I believe in taking care of others and building strong relationships',
            'q2': 'I like having routines and traditions that bring stability',
            'q3': 'I find fulfillment in helping others and making a difference'
        }
    },
    {
        'user_id': 'user_004',
        'name': 'David',
        'mbti_type': 'ENTP',
        'astrology_sign': 'Aquarius',
        'interests': ['technology', 'debate', 'innovation', 'entrepreneurship'],
        'communication_style': 'Direct',
        'prompt_answers': {
            'q1': 'I enjoy intellectual debates and exploring new ideas',
            'q2': 'I thrive on change and new challenges',
            'q3': 'I like to question everything and find innovative solutions'
        }
    }
]

print(f"👥 Created {len(sample_users)} sample user profiles")

# Test compatibility between all pairs
compatibility_results = []

for i, user1 in enumerate(sample_users):
    for j, user2 in enumerate(sample_users):
        if i < j:  # Avoid duplicate pairs
            scores = compatibility_scorer.calculate_overall_compatibility(user1, user2)
            
            result = {
                'user1_id': user1['user_id'],
                'user1_name': user1['name'],
                'user1_mbti': user1['mbti_type'],
                'user2_id': user2['user_id'],
                'user2_name': user2['name'],
                'user2_mbti': user2['mbti_type'],
                **{f'{factor}_score': score for factor, score in scores.items()}
            }
            compatibility_results.append(result)

# Create DataFrame for analysis
compatibility_df = pd.DataFrame(compatibility_results)

print("\n💕 Compatibility Analysis Results:")
display(compatibility_df[['user1_name', 'user2_name', 'user1_mbti', 'user2_mbti', 
                         'mbti_score', 'prompts_score', 'interests_score', 'overall_score']].round(3))
```
